```python
# Cell 2: Consolidated status inventory
# Pulls a single, accurate picture of what exists across Phases 1-6b before
# any synthesis writing happens, so the README and capstone content are
# grounded in an actual inventory, not memory of what each phase claimed.

from google.colab import drive
drive.mount('/content/drive')

import os, json

DRIVE_PATH = "/content/drive/MyDrive/python-ai-governance-p2/data/"

ALL_PHASE_FILES = {
    "Phase 1 (baseline)": "phase01_baseline_documents.json",
    "Phase 2a (RAGAS, Gemini judge)": "phase02a_gemini_judge_results.json",
    "Phase 2b (RAGAS, Claude judge)": "phase02b_claude_judge_results.json",
    "Phase 3a (DeepEval RAG)": "phase03a_deepeval_rag_results.json",
    "Phase 3b (DeepEval governance)": "phase03b_governance_metrics_results.json",
    "Phase 4a (AspectCritic)": "phase04a_aspect_critic_results.json",
    "Phase 4b (judge alignment)": "phase04b_judge_alignment_results.json",
    "Phase 5a (OWASP LLM Top 10)": "phase05a_promptfoo_owasp_results.json",
    "Phase 5b (OWASP Agentic Top 10)": "phase05b_promptfoo_owasp_agentic_results.json",
    "Phase 5c (MITRE ATLAS mapping)": "phase05c_mitre_atlas_mapping_results.json",
    "Phase 6a (Langfuse wiring)": "phase06a_langfuse_custom_scores_results.json",
    "Phase 6b (regression alarm)": "phase06b_regression_alarm_results.json",
}

inventory = {}
for label, filename in ALL_PHASE_FILES.items():
    path = DRIVE_PATH + filename
    if os.path.exists(path):
        with open(path) as f:
            data = json.load(f)
        if isinstance(data, list):
            # Phase 1's baseline documents file: a raw list, not a results
            # dict, no "simulated" concept applies to it at all.
            status = "N/A (raw data file, not an evaluation result)"
        else:
            status = data.get("simulated", "no field (Phase 5c/6a are mixed by design)")
        inventory[label] = {"found": True, "status": status}
    else:
        inventory[label] = {"found": False, "status": "MISSING"}

print("PROJECT 2 CONSOLIDATED STATUS")
print("=" * 70)
for label, info in inventory.items():
    if not info["found"]:
        flag = "MISSING"
    elif info["status"] is True:
        flag = "SIMULATED"
    elif info["status"] is False:
        flag = "REAL"
    else:
        flag = "MIXED/N-A"
    print(f"  {label}: {flag}")

found_count = sum(1 for v in inventory.values() if v["found"])
print()
print(f"Phases found: {found_count}/{len(ALL_PHASE_FILES)}")
print("Every phase's evaluation LOGIC is real, verified, working code.")
print("Zero phases have executed a real live evaluation run yet.")
```
